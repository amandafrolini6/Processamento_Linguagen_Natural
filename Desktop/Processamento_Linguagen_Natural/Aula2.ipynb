```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np
```

### Exercício 1

Considere o seguinte conjunto de documentos (_corpus_):
- doc1 = `o gato dorme no sofá`
- doc2 = `o cachorro dorme no tapete`
- doc3 = `o gato caça o rato`
- doc4 = `o cachorro corre no parque`

Utilizando TF-IDF, faça o seguinte:
- Crie uma lista com os documentos.
- Utilize `TfidfVectorizer` para transformar os textos em uma matriz TF-IDF.
- Exiba: a matriz resultante e as palavras (features) extraídas do corpus (_vocabulario_)
- Identifique manualmente quais palavras possuem maior peso em cada documento
- Explique por que palavras como "o" possuem peso baixo. 

```python
doc1 = "o gato dorme no sofá"
doc2 = "o cachorro dorme no tapete"
doc3 = "o gato caça o rato"
doc4 = "o cachorro corre no parque"

docs = [doc1, doc2, doc3, doc4]

vectorizer = TfidfVectorizer()

tfidf = vectorizer.fit_transform(docs)

print("Palavras (features):")
print(vectorizer.get_feature_names_out())

print("\nMatriz TF-IDF:")
print(tfidf.toarray())
```

### Exercicio 2

Considere os documentos abaixo sobre tecnologia:
- doc1 = `python é muito usado para ciência de dados`
- doc2 = `java é muito usado no desenvolvimento corporativo`
- doc3 = `python é popular em machine learning`
- doc4 = `javascript é usado no desenvolvimento web`

Tarefas:
- Gere a matriz TF-IDF.
- Para cada documento, encontre: as 3 palavras com maior peso TF-IDF.

```python
docs = [
    "python é muito usado para ciência de dados",
    "java é muito usado no desenvolvimento corporativo",
    "python é popular em machine learning",
    "javascript é usado no desenvolvimento web"
]

vectorizer = TfidfVectorizer()

tfidf = vectorizer.fit_transform(docs)

features = vectorizer.get_feature_names_out()

matriz = tfidf.toarray()

for i, doc in enumerate(matriz):

    top_indices = doc.argsort()[-3:][::-1]

    palavras = [features[j] for j in top_indices]

    print(f"Documento {i+1} -> {palavras}")
```

### Exercicio 3

Considere o conjunto de frases (corpus):
- doc1 = `o banco aprovou o empréstimo do cliente`
- doc2 = `o cliente solicitou um empréstimo no banco`
- doc3 = `o jogador marcou um gol no campeonato`
- doc4 = `o time venceu o campeonato de futebol`

Tarefas:
- Construa a matriz TF-IDF.
- Calcule a similaridade de cosseno entre todos os documentos.
- Identifique: quais documentos são mais semelhantes e quais são mais diferentes
- Discussão: Por que os documentos sobre banco e empréstimo são mais similares entre si do que com os documentos sobre futebol?


```python
docs = [
    "o banco aprovou o empréstimo do cliente",
    "o cliente solicitou um empréstimo no banco",
    "o jogador marcou um gol no campeonato",
    "o time venceu o campeonato de futebol"
]

vectorizer = TfidfVectorizer()

tfidf = vectorizer.fit_transform(docs)

similaridade = cosine_similarity(tfidf)

print("Matriz de Similaridade:")
print(similaridade)
```

### Exercicio 4

Considere o seguinte conjunto de documentos:
- doc1 = `machine learning é usado para previsão de vendas`
- doc2 = `redes neurais são usadas em reconhecimento de imagem`
- doc3 = `análise de dados ajuda empresas a tomar decisões`
- doc4 = `algoritmos de aprendizado de máquina analisam grandes volumes de dados`

Agora considere a consulta do usuário:
- query = `aprendizado de máquina e dados`

Tarefas:
- Treine um vetor TF-IDF com os documentos.
- Transforme a consulta em vetor TF-IDF.
- Calcule a similaridade entre a consulta e cada documento.
- Exiba os documentos ordenados do mais relevante para o menos relevante.

```python
docs = [
    "machine learning é usado para previsão de vendas",
    "redes neurais são usadas em reconhecimento de imagem",
    "análise de dados ajuda empresas a tomar decisões",
    "algoritmos de aprendizado de máquina analisam grandes volumes de dados"
]

query = ["aprendizado de máquina e dados"]

vectorizer = TfidfVectorizer()

tfidf_docs = vectorizer.fit_transform(docs)

tfidf_query = vectorizer.transform(query)

similaridades = cosine_similarity(tfidf_query, tfidf_docs)

print("Similaridades:", similaridades)

indices = np.argsort(similaridades[0])[::-1]

print("\nDocumentos ordenados por relevância:")

for i in indices:
    print(docs[i])
```
