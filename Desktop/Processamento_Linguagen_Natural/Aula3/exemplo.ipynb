```python

from gensim.models import Word2Vec

sentences = [
    ["gato", "come", "peixe"],
    ["cachorro", "come", "carne"],
    ["gato", "e", "cachorro", "são", "animais"]
]

# Treinando o modelo
# vector_size: tamanho do vetor (embedding)
# window: janela de contexto
# min_count: ignora palavras que aparecem menos que x vezes
model = Word2Vec(sentences, vector_size=5, window=2, min_count=1)

print(model.wv['gato'])  # Vetor de embedding para a palavra "gato"
```

Exercícios

Dada uma lista de frases (corpus), utilize o Word2Vec do gensim para treinar um modelo com uma dimensão de vetor (vector_size) igual a 100 e uma janela (window) de 5.

```python
from gensim.models import Word2Vec

sentencas = [
    ["o", "processamento", "de", "linguagem", "natural", "e", "incrível"],
    ["o", "aprendizado", "de", "máquina", "e", "uma", "subarea", "da", "ia"],
    ["redes", "neurais", "sao", "usadas", "em", "deep", "learning"],
    ["o", "gensim", "e", "excelente", "para", "word", "embeddings"]
]

modelo = Word2Vec(sentences=sentencas, vector_size=100, window=5, min_count=1, workers=4)
print("Modelo treinado com sucesso.")
```

Use o modelo treinado no Exercício 1 (ou um modelo pré-treinado) para encontrar as 3 palavras mais similares à palavra "linguagem

```python
try:

    similares = modelo.wv.most_similar("linguagem", topn=3)

    print("Palavras similares a 'linguagem':")
    
    for palavra, score in similares:
        print(f"- {palavra}: {score:.4f}")

except KeyError:
    print("Palavra não encontrada no vocabulário.")
```
